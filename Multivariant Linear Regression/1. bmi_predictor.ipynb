```python
"""Import required libraries for data manipulation,
plotting, and modeling."""
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn import linear_model
```

```python
"""Load the Tumor Growth Rate prediction dataset
from Google Drive."""
df= pd.read_excel("/content/drive/MyDrive/ML-Model Training/Multivariant Linear Regression/Weight-BMI Predictive Dataset (50 Samples) .xlsx")
df
```

```python
 #Initialize and train the Multivariant Linear Regression model
lr_tgr=linear_model.LinearRegression()
lr_tgr.fit(df[['Height','Weight']],df['BMI'])
```

```python
"""Test the model by predicting the 'BMI'
for a specific 'Height'and 'Weight'."""
lr_tgr.predict([[3.654,94.757575]])
```

```python
# Calculate the R-squared score to measure model accuracy
R2_Score=lr_tgr.score(df[['Height','Weight']],df['BMI'])
print(R2_Score)
```

```python
# Retrieve the trained model's coefficient (slope / weight)
lr_tgr.coef_
```

```python
# Retrieve the trained model's Y-intercept (bias)
lr_tgr.intercept_
```
